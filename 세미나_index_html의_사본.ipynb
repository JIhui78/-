## 3. Python 백엔드 구조화 (FastAPI) 및 데이터 모델 정의

`localStorage`에 저장되던 데이터 구조를 바탕으로 `pydantic`을 사용하여 FastAPI에서 사용할 데이터 모델을 정의합니다. 이 모델들은 Supabase 테이블의 스키마와 일치해야 합니다.

**Supabase에 필요한 테이블 스키마 (예시):**

**`users` 테이블:**

*   `id`: `TEXT` (Primary Key, 사용자 학번)
*   `pw`: `TEXT`
*   `name`: `TEXT`
*   `dept`: `TEXT`
*   `msg`: `TEXT` (기본값: '열공 중!')
*   `img`: `TEXT` (기본값: '기본 이미지 URL')
*   `created_at`: `TIMESTAMP WITH TIME ZONE` (기본값: `now()`)

**`study_logs` 테이블:**

*   `log_id`: `UUID` (Primary Key, 자동으로 생성)
*   `user_id`: `TEXT` (외래 키, `users.id` 참조)
*   `date`: `DATE`
*   `seconds`: `INT`
*   `created_at`: `TIMESTAMP WITH TIME ZONE` (기본값: `now()`)
*   `CONSTRAINT`: `user_id`, `date` 조합이 Unique (사용자별 일일 학습 기록)

아래 코드는 FastAPI 애플리케이션의 기본 구조와 `User`, `StudyLog` 모델 정의 예시입니다. 다른 데이터 (이벤트, 카테고리, 게시글, 댓글)에 대해서도 유사하게 모델을 정의해야 합니다.

```python
from fastapi import FastAPI, HTTPException
from fastapi.responses import HTMLResponse
from fastapi.staticfiles import StaticFiles
from pydantic import BaseModel
from typing import List, Optional
from datetime import date, datetime
import uvicorn
import os

# FastAPI 애플리케이션 초기화
app = FastAPI()

# 'static' 디렉토리가 없으면 생성
if not os.path.exists('static'):
    os.makedirs('static')

# index.html 파일을 'static' 디렉토리로 이동 (FastAPI가 StaticFiles로 서빙할 수 있도록)
# 이 코드는 Colab 환경에서 파일을 옮기는 예시이며, 실제 프로젝트에서는 'static' 폴더에 직접 HTML 파일을 두게 됩니다.
# Colab의 첫 번째 셀에서 index.html을 생성하므로, 여기서는 해당 파일을 'static'으로 복사하거나 심볼릭 링크를 걸 수도 있습니다.
# 편의를 위해, 직접 HTMLResponse로 메인 페이지를 서빙하고, 다른 정적 파일은 StaticFiles로 서빙하겠습니다.

# Pydantic 모델 정의 (Supabase 테이블 스키마와 매핑)

class UserBase(BaseModel):
    id: str
    pw: str
    name: str
    dept: str
    msg: Optional[str] = '열공 중!'
    img: Optional[str] = 'https://cdn-icons-png.flaticon.com/512/3135/3135715.png' # 기본 이미지 URL

class UserCreate(UserBase):
    pass

class UserInDB(UserBase):
    created_at: datetime = None

    class Config:
        from_attributes = True


class StudyLogBase(BaseModel):
    user_id: str
    date: date
    seconds: int

class StudyLogCreate(StudyLogBase):
    pass

class StudyLogInDB(StudyLogBase):
    log_id: str # Supabase에서 생성될 UUID
    created_at: datetime = None

    class Config:
        from_attributes = True


# --- API 엔드포인트 예시 ---

# 사용자 회원가입
@app.post("/users/signup", response_model=UserInDB)
async def signup_user(user: UserCreate):
    try:
        # Supabase에 사용자 데이터 삽입
        # 비밀번호는 실제 서비스에서는 해싱하여 저장해야 합니다.
        response = supabase.table('users').insert(user.model_dump()).execute()
        if response.data:
            return UserInDB(**response.data[0])
        raise HTTPException(status_code=400, detail="사용자 등록 실패")
    except Exception as e:
        # 이미 존재하는 ID (Primary Key 충돌) 등 에러 처리
        if "duplicate key value" in str(e):
            raise HTTPException(status_code=400, detail="이미 존재하는 학번입니다.")
        raise HTTPException(status_code=500, detail=str(e))

# 사용자 로그인
@app.post("/users/login")
async def login_user(user: UserBase):
    try:
        response = supabase.table('users').select("*", count='exact').eq('id', user.id).eq('pw', user.pw).execute()
        if response.data and len(response.data) > 0:
            # 실제 앱에서는 JWT 토큰 등을 발행하여 인증 처리
            return {"message": "로그인 성공", "user": UserInDB(**response.data[0])}
        raise HTTPException(status_code=401, detail="학번 또는 비밀번호가 틀립니다!")
    except Exception as e:
        raise HTTPException(status_code=500, detail=str(e))

# 학습 기록 추가 또는 업데이트
@app.post("/study_logs", response_model=StudyLogInDB)
async def upsert_study_log(log: StudyLogCreate):
    try:
        # user_id와 date 기준으로 기존 로그가 있으면 업데이트, 없으면 삽입 (upsert)
        # Supabase Python 클라이언트의 upsert는 현재 PK 기준으로만 동작.
        # 특정 컬럼 조합으로 upsert하려면 SQL 쿼리 또는 별도 로직 필요
        # 여기서는 단순화를 위해 insert_or_update 로직을 가정합니다.

        # 먼저 기존 로그 확인
        existing_log_res = supabase.table('study_logs')\
            .select('*')\
            .eq('user_id', log.user_id)\
            .eq('date', log.date.isoformat())\
            .execute()

        if existing_log_res.data:
            # 기존 로그가 있으면 업데이트
            updated_log = supabase.table('study_logs')\
                .update({'seconds': log.seconds})\
                .eq('user_id', log.user_id)\
                .eq('date', log.date.isoformat())\
                .execute()
            return StudyLogInDB(**updated_log.data[0])
        else:
            # 기존 로그가 없으면 삽입
            new_log_res = supabase.table('study_logs').insert(log.model_dump()).execute()
            return StudyLogInDB(**new_log_res.data[0])

    except Exception as e:
        raise HTTPException(status_code=500, detail=str(e))

# 루트 경로 ('/')에서 index.html을 서빙
@app.get("/", response_class=HTMLResponse)
async def read_root():
    with open("index.html", "r", encoding="utf-8") as f:
        html_content = f.read()
    return HTMLResponse(content=html_content)

# 기타 정적 파일 (CSS, JS, 이미지 등)을 /static 경로를 통해 서빙
# 실제 Vercel 배포 시에는 빌드 설정에서 정적 파일을 처리하는 방식이 더 효율적일 수 있습니다.
# app.mount("/static", StaticFiles(directory="static"), name="static")


# 백엔드 서버를 Colab에서 실행하기 위한 코드 (실제 Vercel 배포 시에는 uvicorn command로 실행)
# 이 셀은 실행하면 서버가 시작되어 다음 셀로 넘어가지 않습니다.
# 개발/테스트 목적으로만 사용하고, 실제 Vercel 배포 시에는 이 부분을 제외합니다.
if __name__ == "__main__":
    # Colab 환경에서는 'localhost' 대신 '0.0.0.0'을 사용해야 외부에서 접근 가능할 수 있습니다.
    # 하지만 Colab 노트북 자체의 네트워크 제한으로 외부 접근이 어려울 수 있습니다.
    # 여기서는 예시로만 제공하며, 실제 Vercel 배포 시에는 Vercel의 빌드 및 실행 과정을 따릅니다.
    print("FastAPI 애플리케이션이 준비되었습니다. 이 셀을 직접 실행하여 서버를 시작할 수 있습니다.\n"\
          "하지만 Colab 환경에서는 외부에서 접근하기 어렵고, 셀이 계속 실행 중 상태가 됩니다.\n"\
          "실제 개발 시에는 로컬 환경에서 실행하거나 Vercel 배포 가이드를 따르세요.")
    # uvicorn.run(app, host="0.0.0.0", port=8000) # 주석 처리하여 셀이 계속 실행되지 않도록 함
```

### `app/crud/user.py` 파일 생성

사용자 관련 데이터베이스 CRUD(Create, Read, Update, Delete) 로직을 `app/crud/user.py` 파일로 분리합니다.

```python
user_crud_content = '''
from fastapi import HTTPException
from app.models.user import UserCreate, UserInDB
from app.core.database import supabase

async def create_user(user: UserCreate) -> UserInDB:
    try:
        response = supabase.table('users').insert(user.model_dump()).execute()
        if response.data:
            return UserInDB(**response.data[0])
        raise HTTPException(status_code=400, detail="User registration failed")
    except Exception as e:
        if "duplicate key value" in str(e):
            raise HTTPException(status_code=400, detail="User ID already exists")
        raise HTTPException(status_code=500, detail=str(e))

async def authenticate_user(user_id: str, password: str) -> Optional[UserInDB]:
    try:
        response = supabase.table('users').select("*", count='exact').eq('id', user_id).eq('pw', password).execute()
        if response.data and len(response.data) > 0:
            return UserInDB(**response.data[0])
        return None
    except Exception as e:
        raise HTTPException(status_code=500, detail=str(e))
'''

with open('app/crud/user.py', 'w', encoding='utf-8') as f:
    f.write(user_crud_content)
print("app/crud/user.py 파일이 생성되었습니다.")
```

```python
user_router_content = '''
from fastapi import APIRouter, HTTPException
from app.models.user import UserCreate, UserInDB, UserBase
from app.crud.user import create_user, authenticate_user

router = APIRouter()

@router.post("/signup", response_model=UserInDB)
async def signup_user_route(user: UserCreate):
    return await create_user(user)

@router.post("/login")
async def login_user_route(user: UserBase):
    authenticated_user = await authenticate_user(user.id, user.pw)
    if not authenticated_user:
        raise HTTPException(status_code=401, detail="Incorrect user ID or password")
    return {"message": "Login successful", "user": authenticated_user}
'''

with open('app/routers/users.py', 'w', encoding='utf-8') as f:
    f.write(user_router_content)
print("app/routers/users.py 파일이 생성되었습니다.")
```
